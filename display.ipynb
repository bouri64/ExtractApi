## Stock price, dividend and stock splits from yfinance

```python
import yfinance as yf
import pandas as pd
companies = df['Symbol'].unique()
concat_yfinance = pd.DataFrame()
for company in companies:
    # 1️⃣ Download full historical data
    try:
        yfinance_df = yf.Ticker(company).history(period="max")
    except:
        print("Issue with", company)
        continue
    # Ensure Date is a column
    yfinance_df = yfinance_df.reset_index()
    yfinance_df['Date'] = pd.to_datetime(yfinance_df['Date'])

    # 2️⃣ Make Date the index for resampling
    yfinance_df.set_index('Date', inplace=True)
    columns_dict = {
        'Dividends': 'sum',       # Sum dividends in the month
        'Stock Splits': 'sum',    # Sum stock splits in the month
        'Close': 'last'           # Closing price at end of month
    }
    valid_columns = {item : columns_dict[item] for item in columns_dict if item in yfinance_df.columns}
    # 3️⃣ Resample monthly
    monthly = yfinance_df.resample('ME').agg(valid_columns)
    if len(valid_columns) < 3:
        print(f"For company {company} only {valid_columns.keys()} are valid")
    if 'Dividends' in monthly.columns:
    # 4️⃣ Calculate rolling 12-month dividend sum
        monthly['Rolling12MDividend'] = monthly['Dividends'].rolling(window=12).sum()
    if 'Stock Splits' in monthly.columns:
        monthly['Rolling12MStockSplit'] = monthly['Stock Splits'].rolling(window=12).sum()
    monthly['Symbol'] = company
    # 5️⃣ Reset index to have Date as column
    monthly = monthly.reset_index()
    concat_yfinance = pd.concat([concat_yfinance,monthly], axis=0)
    # print(monthly.tail(15))
```

```python
# # Remove timezone from yfinance_df
concat_yfinance['Date'] = pd.to_datetime(concat_yfinance['Date']).dt.tz_localize(None)

# Make sure monthly MonthEnd is also naive
df['MonthEnd'] = df['Date'] + pd.offsets.MonthEnd(0)
df['MonthEnd'] = df['MonthEnd'].dt.tz_localize(None)

df = df.merge(concat_yfinance, right_on=['Date','Symbol'], left_on=['MonthEnd','Symbol'], how='left')
```

```python
df = df.rename(columns={"Date_x": "Date"}).drop(columns=["Date_y"])
```

```python
import numpy as np
# Make sure no NaN
df['Rolling12MStockSplit'] = df['Rolling12MStockSplit'].fillna(0)

# Replace 0 with 1 for multiplication
df['Rolling12MStockSplit'] = df['Rolling12MStockSplit'].replace(0, 1)
df = df.sort_values(['Symbol', 'Date'], ascending=[True, False])
df['cumulative_split_factor'] = df.groupby('Symbol')['Rolling12MStockSplit'].cumprod().shift(1, fill_value=1)
```

## Whether or not in sp500

```python
start_end = pd.read_csv("sp500_ticker_start_end.csv", parse_dates=["end_date", "start_date"])
# For each year count
# > start and <= end or end is null
```

```python
def active_sp500(x):
    comp = start_end[start_end['ticker'] == x['Symbol']]
    comp = comp[(comp['end_date']>= x['Date']) + (comp['end_date'].isna())]
    comp = comp[comp['start_date'] < x['Date']]
    return len(comp) > 0
```

```python
df['PartOfSp500'] = df.apply(active_sp500, axis=1)
```

```python
start_end = start_end.fillna(start_end['end_date'].max())
```

```python
def duration_sp500(x):
    comp = start_end[start_end['ticker'] == x['Symbol']]
    time = 0
    for _,c in comp.iterrows():
        time += (c.end_date - c.start_date).days
    return int(time/362.25)
```

```python
df['DurationSp500'] = df.apply(duration_sp500, axis=1)
```
