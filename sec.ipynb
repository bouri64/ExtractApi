```python
columns = ['Year','Symbol', 'Date','Eps','EpsDiluted', 'Assets', 'Liabilities', 'LongTermDebt', 'Equity', 'LiabilityEquity', 'PaymentDividend','CommonDividendDeclared','CommonDividend']
metrics = ['us-gaap:EarningsPerShareBasic', 'us-gaap:EarningsPerShareDiluted', 'us-gaap:Assets', 'us-gaap:Liabilities', 'us-gaap:LongTermDebt', 'us-gaap:StockholdersEquity', 'us-gaap:LiabilitiesAndStockholdersEquity', 'us-gaap:PaymentsOfDividends', 'us-gaap:CommonStockDividendsPerShareDeclared', 'us-gaap:DividendsCommonStock']
```

```python
import pandas as pd
df = pd.read_csv("sp500_ticker_start_end.csv")
```

```python
companies = list(df[(df['end_date']>= '2010-01-01') + (df['end_date'].isna())]['ticker'])
```

```python
def get_value(xbrl, regex, year):
    # Add by_fiscal_year here
    values = xbrl.facts.query().by_concept(regex, exact=True).to_dataframe()
    value = None
    for _,v in values.iterrows():
        if 'period_end' in v.keys():
            f_year = int(v['period_end'][0:4])
        else:
            f_year = int(v['period_instant'][0:4])
        if f_year == year :
            value = v['value']
            return value
    return value
```

```python
data = []
skip_companies = set()
```
